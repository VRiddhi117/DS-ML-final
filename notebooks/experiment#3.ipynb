```python
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.feature_extraction.text import TfidfVectorizer

def log_votes(arr):
    arr = arr.copy()
    arr[:, 1] = np.log1p(arr[:, 1])  # helpful_votes
    return arr

num_pipe = Pipeline([
    ("log_votes", FunctionTransformer(log_votes, validate=True)),
    ("scaler", StandardScaler())
])

preprocessor = ColumnTransformer([
    ("text", TfidfVectorizer(max_features=20000, stop_words="english", ngram_range=(1,2)), "review_text"),
    ("num", num_pipe, ["rating", "helpful_votes"]),
    ("verified", "passthrough", ["verified_purchase"]),
], sparse_threshold=0.3)
```

```python
import pandas as pd

df = pd.read_sql("""
SELECT review_text, rating, helpful_votes, verified_purchase, label
FROM review
""", engine)

df["verified_purchase"] = df["verified_purchase"].astype(int)
df["helpful_votes"] = df["helpful_votes"].fillna(0).astype(int)
df["rating"] = df["rating"].astype(float)
df["label"] = df["label"].astype(int)

X = df[["review_text", "rating", "helpful_votes", "verified_purchase"]].copy()
y = df["label"]

print(df.shape)
print(y.value_counts(normalize=True))
```

```python
import numpy as np

def add_features(df_in: pd.DataFrame) -> pd.DataFrame:
    out = df_in.copy()

    text = out["review_text"].fillna("").astype(str)

    out["review_len"] = text.str.len()
    out["word_count"] = text.str.split().str.len()
    out["exclamation_count"] = text.str.count("!")
    out["caps_ratio"] = text.apply(lambda s: sum(c.isupper() for c in s) / max(len(s), 1))

    out["helpful_log"] = np.log1p(out["helpful_votes"])
    out["rating_x_verified"] = out["rating"] * out["verified_purchase"]

    return out

X_fe = add_features(X)

print(X_fe.head())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

TEXT_COL = "review_text"

NUM_COLS_FE = [
    "rating", "helpful_votes", "verified_purchase",
    "review_len", "word_count", "exclamation_count", "caps_ratio",
    "helpful_log", "rating_x_verified"
]

pre = ColumnTransformer(
    transformers=[
        ("text", TfidfVectorizer(max_features=5000, stop_words="english"), TEXT_COL),
        ("num", StandardScaler(with_mean=False), NUM_COLS_FE),
    ]
)

pipe = Pipeline([
    ("pre", pre),
    ("clf", LogisticRegression(max_iter=2000, solver="liblinear"))
])

pipe
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.metrics import f1_score, confusion_matrix

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(pipe, X_fe, y, cv=cv, scoring="f1")
f1_mean = float(scores.mean())
f1_std = float(scores.std())

preds = cross_val_predict(pipe, X_fe, y, cv=cv)
tn, fp, fn, tp = confusion_matrix(y, preds).ravel()

pipe.fit(X_fe, y)
train_pred = pipe.predict(X_fe)
f1_train = float(f1_score(y, train_pred))

run_name = "EXP3_feature_engineering"

with mlflow.start_run(run_name=run_name):
    mlflow.log_param("experiment", "exp3")
    mlflow.log_param("model", "logreg")
    mlflow.log_param("feature_engineering", True)
    mlflow.log_param("tfidf_max_features", 5000)
    mlflow.log_param("folds", 5)

    mlflow.log_metric("f1_cv_mean", f1_mean)
    mlflow.log_metric("f1_cv_std", f1_std)
    mlflow.log_metric("f1_train", f1_train)

    mlflow.log_metric("tp", int(tp))
    mlflow.log_metric("tn", int(tn))
    mlflow.log_metric("fp", int(fp))
    mlflow.log_metric("fn", int(fn))

print(f"✅ {run_name}: f1_mean={f1_mean:.4f} std={f1_std:.4f} train={f1_train:.4f}")
print("TP TN FP FN:", tp, tn, fp, fn)
```
